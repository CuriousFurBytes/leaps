## Part 2: Python as a Calculator

Python can do arithmetic. The operators are:

| Operator | Meaning | Example |
|----------|---------|--------|
| `+` | Addition | `3 + 2` → `5` |
| `-` | Subtraction | `10 - 4` → `6` |
| `*` | Multiplication | `3 * 7` → `21` |
| `/` | True division | `7 / 2` → `3.5` |
| `//` | Floor division | `7 // 2` → `3` |
| `%` | Modulo (remainder) | `7 % 2` → `1` |
| `**` | Exponentiation | `2 ** 8` → `256` |

````python
# Basic arithmetic
print(2 + 3)      # Addition: 5
print(10 - 4)     # Subtraction: 6
print(3 * 7)      # Multiplication: 21
print(15 / 4)     # True division: 3.75 (always float in Python 3)
print(15 // 4)    # Floor division: 3 (rounds down)
print(15 % 4)     # Modulo: 3 (remainder of 15 ÷ 4)
print(2 ** 10)    # Exponentiation: 1024
````

````python
# Order of operations — same as mathematics (PEMDAS)
print(2 + 3 * 4)       # 14 — multiplication first
print((2 + 3) * 4)     # 20 — parentheses first
print(2 ** 3 ** 2)     # 512 — ** is right-associative: 2**(3**2) = 2**9 = 512

# Python integers have UNLIMITED size — no overflow!
print(2 ** 100)    # 1267650600228229401496703205376 — a 31-digit number!
````

````python
# The math module gives us more functions
import math

print(math.pi)         # 3.141592653589793
print(math.e)          # 2.718281828459045 (Euler's number)
print(math.sqrt(144))  # 12.0
print(math.floor(3.7)) # 3 — round down
print(math.ceil(3.2))  # 4 — round up
print(math.factorial(10))  # 3628800
````

## Part 3: Variables — Giving Names to Values

A **variable** is a name that refers to a value. You create one with the `=` assignment operator.

Think of variables as **labels** or **sticky notes** attached to values — not as boxes containing values.

````python
# Variable assignment
name = "Alice"      # str (string)
age = 30            # int (integer)
height = 1.75       # float (decimal number)
is_student = False  # bool (True or False)

print(name)
print(age)
print(height)
print(is_student)
````

````python
# Variables can be used in calculations
years_until_retirement = 65 - age
print(f"Years until retirement: {years_until_retirement}")

# Variables can be updated
age = age + 1      # one year later
# or equivalently: age += 1
print(f"Next year, age will be: {age}")
````

````python
# Python is dynamically typed — a variable can hold any type
x = 42
print(f"x = {x}, type = {type(x).__name__}")

x = "now a string"  # reassigning to a different type
print(f"x = {x!r}, type = {type(x).__name__}")

x = [1, 2, 3]       # now a list
print(f"x = {x}, type = {type(x).__name__}")
````

### f-strings: Embedding Expressions in Strings

f-strings (formatted string literals) let you embed Python expressions directly inside strings. Put `f` before the opening quote, and use `{}` to embed expressions.

````python
name = "Alice"
age = 30
score = 9.5

# Basic f-string
print(f"Hello, {name}!")

# Expressions inside {}
print(f"In 10 years, {name} will be {age + 10} years old.")

# Format specifiers after :
print(f"Score: {score:.1f}%")          # 1 decimal place
print(f"Score: {score:.0f}%")          # 0 decimal places (rounded)
print(f"Age: {age:05d}")               # 5 digits, zero-padded
print(f"{1000000:,}")                  # Thousands separator

# Debug format (Python 3.8+)
print(f"{name=}")                      # name='Alice'
print(f"{age=}")                       # age=30

# Today's date in an f-string
import datetime
today = datetime.date.today()
print(f"Hello, {name}! Today is an excellent day to learn Python.")
print(f"Date: {today:%B %d, %Y}")
````

## Part 4: The Zen of Python

Python has a built-in Easter egg: a collection of guiding principles called the **Zen of Python** (PEP 20). Run the cell below to see it.

````python
import this
````

### Three Principles Worth Understanding

**1. "Readability counts"**

Python code is designed to be read as much as to be executed. When you write code, write it for the human reader — not just the computer. Use descriptive names, clear structure, and comments when needed.

```python
# Hard to read:
def f(x):
    return x*2+1

# Easy to read:
def double_plus_one(x):
    return x * 2 + 1
```

**2. "Simple is better than complex"**

Don't reach for a complex solution when a simple one will do. Python rewards simplicity — the idiomatic "Pythonic" approach is usually the readable, simple one.

**3. "Errors should never pass silently"**

When something goes wrong, Python should say so. Don't catch and ignore errors. A program that hides its failures is much harder to debug than one that fails loudly with a clear error message.

````python
# Demonstrating "Readability counts"

# Less readable (but valid Python)
x=[i**2 for i in range(1,11) if i%2==0]
print(x)

# More readable (same result)
even_squares = [n**2 for n in range(1, 11) if n % 2 == 0]
print(even_squares)
````
